```python
import cogent3
from cogent3 import get_app
from cogent3 import load_aligned_seqs
import matplotlib.pyplot as plt
import paths
import os

def singlentmodel_cds():
    GN_subsmodel = get_app("model", "GN", time_het="max", lf_args={"discrete_edges": ["Orangutan"]}, optimise_motif_probs=False, show_progress=False)
    return GN_subsmodel

def singlentmodel_noncds():
    GN_subsmodel = get_app("model", "GN", time_het="max", lf_args={"discrete_edges": ["Orangutan"]}, optimise_motif_probs=False, show_progress=False)
    return GN_subsmodel
```

```python
region_name = "intergenicAR"
chromosome_name = "11"

region = region_name + "/chrm" + chromosome_name
folder_in = paths.DATA_HUMCHIMPORANG115 + region
file_in = folder_in + "/filtered.fa"
alns = load_aligned_seqs(file_in, moltype="dna")

data_out = folder_in + "/sm_output"
os.makedirs(data_out, exist_ok=True)

print("alignment length:", len(alns)/10**9)
```

```python
omit_gap_pos = get_app("omit_degenerates", moltype="dna", motif_length=1)
aln_gapped = omit_gap_pos(alns)

print("alignment length:", len(aln_gapped)/10**6)
```

```python
if region_name == "cds":
    sm = singlentmodel_cds()
    result_sm = sm(aln_gapped)
else:
    sm = singlentmodel_noncds()
    result_sm = sm(aln_gapped)

result_sm.lf
```
