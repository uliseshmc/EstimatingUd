```python
import cogent3
from cogent3 import get_app
from cogent3 import load_aligned_seqs
from cogent3 import load_unaligned_seqs
from cogent3.app.typing import AlignedSeqsType
from cogent3.app.typing import UnalignedSeqsType
from cogent3.app.typing import SeqType
import paths
import libs
```

```python
def gethumanseq_noncds_aligned(seqs: AlignedSeqsType) -> SeqType:
    """
    A function to get human sequence for noncds aligned sequences collection.
    """
    name_map = {
        "homo_sapiens": "Human"
    }

    seqs = seqs.renamed_seqs(lambda x: name_map.get(x.split(":")[0], x))
    humanaln = seqs.take_seqs(list(name_map.values()))
    humanseq = humanaln.seqs["Human"]
    humanseq = humanseq.seq
    
    return humanseq
```

```python
folder_in = paths.DATA_HUMCHIMPORANGOR114 + 'intronsAR/alldata_chrm22/'

sequence = 'ENSG00000025770.fa'

aln = load_aligned_seqs(filename = folder_in + sequence, moltype='dna')
aln
```

```python
gethumanseq = libs.gethumanseq_noncds_aligned()
humanseq = gethumanseq(aln)
humanseq
```

```python
def gethumanseqs_noncds():
    loader = get_app("load_aligned", moltype="dna")
    get_human_seq = libs.gethumanseq_noncds_aligned(motif_length=3)
    
    noncds_app = loader + get_human_seq

    return noncds_app

folder_in = paths.DATA_HUMCHIMPORANGOR114 + 'intronsAR/alldata_chrm22/'

in_dstore = cogent3.open_data_store(folder_in, suffix='fa', mode='r')

noncds_app = gethumanseqs_noncds()
#r.obj saves the sequence; r.source saves the file that was used
nonconcat_noncds = [r.obj for r in noncds_app.as_completed(in_dstore[:20], parallel=False) if r]
concat_noncds = cogent3.make_seq("".join(str(s) for s in nonconcat_noncds),name="Human",moltype="dna")
concat_noncds
```
